```python
# fetch_itc_prices.py
# Fetches ITC daily adjusted close prices from 1996 to 2026
# Output columns: Date, Adj Close

import yfinance as yf
import pandas as pd

ticker = "ITC.NS"
start_date = "1996-01-01"
end_date = "2026-04-30"

raw = yf.download(ticker, start=start_date, end=end_date, auto_adjust=True)

df = raw[["Close"]].copy()
df.index.name = "Date"
df.columns = ["Adj Close"]
df.reset_index(inplace=True)

df.to_csv("itc_daily_prices_1996_2026.csv", index=False)
print(f"Saved: {len(df)} rows")
```
