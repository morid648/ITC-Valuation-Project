```python
# fetch_peer_beta_data.py
# Fetches peer company market cap, debt, beta for WACC calculation
# Output columns: Name, Ticker, CMP (Rs.), Mar Cap (Rs.Cr.), Debt (Rs.Cr.),
#                 Debt/Eq, Beta

import yfinance as yf
import pandas as pd

peers = {
    "Hind. Unilever"  : "HINDUNILVR.NS",
    "ITC"             : "ITC.NS",
    "Nestle India"    : "NESTLEIND.NS",
    "Varun Beverages" : "VBL.NS",
    "Britannia Inds." : "BRITANNIA.NS",
}

records = []

for name, ticker in peers.items():
    stock = yf.Ticker(ticker)
    info  = stock.info

    cmp     = info.get("currentPrice",          None)
    mkt_cap = info.get("marketCap",             None)
    debt    = info.get("totalDebt",             None)
    beta    = info.get("beta",                  None)

    # Convert market cap and debt from absolute to Rs. Cr.
    mkt_cap_cr = round(mkt_cap / 1e7, 2) if mkt_cap else None
    debt_cr    = round(debt    / 1e7, 2) if debt    else None
    debt_eq    = round(debt_cr / mkt_cap_cr, 2) if (debt_cr and mkt_cap_cr) else None

    records.append({
        "Name"              : name,
        "Ticker"            : ticker,
        "CMP (Rs.)"         : cmp,
        "Mar Cap (Rs.Cr.)"  : mkt_cap_cr,
        "Debt (Rs.Cr.)"     : debt_cr,
        "Debt / Eq"         : debt_eq,
        "Beta"              : beta,
    })

df = pd.DataFrame(records)
df.to_csv("peer_beta_fundamentals.csv", index=False)
print(df.to_string(index=False))
```
